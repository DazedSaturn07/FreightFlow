# Inspect the source
A source row includes foreign-route fields and a distance band. Every row is counted; detailed sampling uses the first 100,000 rows. Read quality.py alongside these cells.

```python
import pandas as pd
from freightflow.config import settings
from freightflow.quality import profile_csv, save_profile, write_manifest
config = settings()
```

```python
pd.read_csv(config.source_csv, nrows=5).loc[:, ["dms_orig", "dms_dest", "dms_mode", "sctg2", "tons_2024"]]
```

```python
report = profile_csv(config.source_csv)
save_profile(report, config.quality_dir / "data_quality_report.json")
write_manifest(config.source_csv, config.quality_dir / "source_manifest.json", report.row_count)
report.row_count, report.column_count, report.duplicate_sample_keys
```
